## 2. Car Speed Check

```python
'''
Question:
Create a Car class with attributes brand and speed.
If speed is greater than 80, print:
"Over Speed"
Otherwise print:
"Safe Driving"
What you need to learn:
• Access object attributes.
• Use if condition.'''
class Car:
    def __init__(self,brand,speed):
        self.brand=brand
        self.speed=speed
    def check_speed(self):
        if self.speed>80:
            print("Over Speed")
        else:
            print("Safe Driving")
car1=Car("BMW",78)
car2=Car("TOYOTO",168.26)

print("Car Brand:",car1.brand)
print("Speed :",car1.speed)
car1.check_speed()
print("\n""------------------------------""\n")
print("Car Brand:",car2.brand)
print("Speed :",car2.speed)
car2.check_speed()
    
```

## 3. Employee Salary Update

```python
'''
Question:
Create an Employee class with attributes name and salary.
Increase salary by 5000 and display the updated salary.
Example:
Salary = 25000
Updated Salary = 30000'''
class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    def update_salary(self):
        self.salary+=5000
emp1=Employee("Sai",25000)
print("Employee Name:",emp1.name)
print("Original Salary:",emp1.salary)

emp1.update_salary()
print("Updated Salary:",emp1.salary)
        
```

## 4. Product Stock Check

```python
''' 
Question:
Create a Product class with attributes:
product_name
stock
If stock is 0:
Out of Stock
Otherwise:
Available 
'''
class Product:
    def __init__(self,product_name,stock):
        self.product_name=product_name
        self.stock=stock
    def ava_stock(self):
        if self.stock<=0:
            print("Out of Stock")
        else:
            print("Available")
obj=Product("Aditya Birla",15)
print("Product Name:",obj.product_name)
print("Stock of the Product:",obj.stock)
obj.ava_stock()
```

## 7. Student Result System

```python
'''
Question:
Create a Student class with:
name
marks
Condition:
marks >= 90 → Grade A
marks >= 75 → Grade B
marks >= 50 → Grade C
otherwise Fail
Concepts:
• Multiple conditions (if-elif-else)
'''
class BankAccount:
    def __init__(self,name,balance):
        self.name=name
        self.balance=balance
    def deposit(self):
        self.balance+=5000

obj=BankAccount("Sai",17500)
print("Account Holder Name:",obj.name)
print("Original Balance:",obj.balance)
obj.deposit()
print("Updated Balance:",obj.balance)

if obj.balance>20000:
    print("Premium Account")
else:
    print("Regular Account")
        
```

## 7. Student Result System

```python
'''
Question:
Create a Student class with:
name
marks
Condition:
marks >= 90 → Grade A
marks >= 75 → Grade B
marks >= 50 → Grade C
otherwise Fail
Concepts:
• Multiple conditions (if-elif-else)
'''
class Student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
    def check_marks(self):
        if self.marks>=90:
            print("Grade-A")
        elif self.marks>=75:
            print("Grade-B")
        elif self.marks>=50:
            print("Grade-c")
        else:
            print("Fail")
obj=Student("Vardhan",88)
print("Student Name:",obj.name)
print("Marks Secured:",obj.marks)
obj.check_marks()
```

## 8. Laptop Price Discount

```python
'''
Question:
Create a Laptop class with:
brand
price
If price is greater than 50000:
Apply 10% discount.
Display updated price.
'''

class Laptop:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    # Method to apply discount
    def apply_discount(self):
        if self.price > 50000:
            self.price = self.price - (self.price * 10 / 100)

laptop1 = Laptop("Dell", 60000)

print("Laptop Brand:", laptop1.brand)
print("Original Price:", laptop1.price)

laptop1.apply_discount()


print("Updated Price:", laptop1.price)
```

 ## 9. Movie Ticket Booking

```python
'''
Question:
Create a MovieTicket class with:

name
age
Condition:
Age >= 18 → Allowed
Age < 18 → Not Allowed
After checking, update age by adding 1 year and display it.
'''
class MovieTicket:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def check_age(self):
        if self.age>=18:
            print("Allowed")
        else:
            print("Not Allowed")
        self.age+=1
obj=MovieTicket("Sai",21)
print("Name:",obj.name)
print("Age:",obj.age)
obj.check_age()
print("Updated age:",obj.age)
            
```

## 10. Online Shopping Cart

```python
'''
Question:
Create a ShoppingCart class with:
customer_name
total_amount
If total amount is greater than ₹1000:
Apply 20% discount
Otherwise:
Apply 5% discount
Display:
• Original Amount
• Discount
• Final Amount
'''
# Creating ShoppingCart class
class ShoppingCart:
    def __init__(self, customer_name, total_amount):
        self.customer_name = customer_name
        self.total_amount = total_amount

    # Method to calculate discount
    def apply_discount(self):
        original_amount = self.total_amount

        if self.total_amount > 1000:
            discount = self.total_amount * 20 / 100
        else:
            discount = self.total_amount * 5 / 100

        final_amount = self.total_amount - discount

        print("Original Amount:", original_amount)
        print("Discount:", discount)
        print("Final Amount:", final_amount)


# Creating object
customer1 = ShoppingCart("Rahul", 2000)

# Display customer name
print("Customer Name:", customer1.customer_name)

# Applying discount
customer1.apply_discount()
```
