# Single Inheritance

## 1. Student Management System (Easy)
### Parent Class
### Student
### Attributes:
#### • student_name
#### • roll_no
### Method:
#### • display_student()
### Child Class
### CollegeStudent
### Attribute:
#### • course_name
### Method:
#### • display_course()
### Requirement:
### Use super().

```python
# Parent Class
class Student:
    def __init__(self, student_name, roll_no):
        self.student_name = student_name
        self.roll_no = roll_no

    def display_student(self):
        print("Student Name :", self.student_name)
        print("Roll Number  :", self.roll_no)


# Child Class
class CollegeStudent(Student):
    def __init__(self, student_name, roll_no, course_name):
        super().__init__(student_name, roll_no)   # Calls parent constructor
        self.course_name = course_name

    def display_course(self):
        self.display_student()    # Calls parent method
        print("Course Name  :", self.course_name)


# Driver Code
student = CollegeStudent("Sai", 101, "Computer Science")

student.display_course()
```

## 2. Vehicle Information (Easy)
### Parent Class
### Vehicle
### Attributes:
#### • brand
#### • model
### Method:
#### • show_vehicle()
### Child Class
### Bike
### Attribute:
#### • mileage
### Method:
#### • show_bike()
### Requirement:
#### Do NOT use super().

```python
# Parent Class
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def show_vehicle(self):
        print("Brand   :", self.brand)
        print("Model   :", self.model)


# Child Class
class Bike(Vehicle):
    def __init__(self, brand, model, mileage):
        # Calling the parent constructor directly (without super())
        Vehicle.__init__(self, brand, model)
        self.mileage = mileage

    def show_bike(self):
        self.show_vehicle()    # Calls parent method
        print("Mileage :", self.mileage, "km/l")


# Driver Code
bike = Bike("Honda", "Shine", 65)
bike.show_bike()
```

```python
# Parent Class
class Employee:
    def __init__(self, employee_name, salary):
        self.employee_name = employee_name
        self.salary = salary

    def display_employee(self):
        print("Employee Name :", self.employee_name)
        print("Salary        : ₹", self.salary)


# Child Class
class Developer(Employee):
    def __init__(self, employee_name, salary, programming_language):
        super().__init__(employee_name, salary)   # Calls parent constructor
        self.programming_language = programming_language

    def display_developer(self):
        self.display_employee()    # Calls parent method
        print("Programming Language :", self.programming_language)


# Driver Code
developer = Developer("Sai", 80000, "Python")
developer.display_developer()
```

## 4. Bank System with Abstraction (Medium)
### Abstract Class
### Bank
### Attributes:
#### • account_holder
#### • account_number
### Abstract Method:
#### • interest_rate()
### Child Class
### SBI
### Implement:
#### • interest_rate()
### Requirement:
#### Use super().

```python
from abc import ABC, abstractmethod

# Abstract Class
class Bank(ABC):
    def __init__(self, account_holder, account_number):
        self.account_holder = account_holder
        self.account_number = account_number

    @abstractmethod
    def interest_rate(self):
        pass


# Child Class
class SBI(Bank):
    def __init__(self, account_holder, account_number):
        super().__init__(account_holder, account_number)   # Calls parent constructor

    def interest_rate(self):
        print("Bank Name      : SBI")
        print("Account Holder :", self.account_holder)
        print("Account Number :", self.account_number)
        print("Interest Rate  : 6.5% per annum")


# Driver Code
customer = SBI("Sai", "123456789")
customer.interest_rate()
```

## 5. Animal Information (Easy)
### Parent Class
### Animal
### Attribute:
#### • animal_name
### Method:
#### • eat()
### Child Class
### Dog
### Method:
#### • bark()
### Requirement:
### No super().

```python
# Parent Class
class Animal:
    def __init__(self, animal_name):
        self.animal_name = animal_name

    def eat(self):
        print("Animal Name :", self.animal_name)
        print(self.animal_name, "is eating.")


# Child Class
class Dog(Animal):
    def __init__(self, animal_name):
        # Calling parent constructor directly (without super())
        Animal.__init__(self, animal_name)

    def bark(self):
        self.eat()      # Calls parent method
        print(self.animal_name, "is barking: Woof! Woof!")


# Driver Code
dog = Dog("German Shepherd")
dog.bark()
```

# Multilevel Inheritance

## 6. School Management System (Easy)
### Grand Parent
### Person
### Attribute:
#### • name
### Parent
### Student
### Attribute:
#### • roll_no
### Child
### CollegeStudent
### Attribute:
#### • course_name
### Display all details.
### Requirement:
#### Use super().

```python
# Grand Parent Class
class Person:
    def __init__(self, name):
        self.name = name


# Parent Class
class Student(Person):
    def __init__(self, name, roll_no):
        super().__init__(name)      # Calls Person constructor
        self.roll_no = roll_no


# Child Class
class CollegeStudent(Student):
    def __init__(self, name, roll_no, course_name):
        super().__init__(name, roll_no)   # Calls Student constructor
        self.course_name = course_name

    def display_details(self):
        print("Name        :", self.name)
        print("Roll Number :", self.roll_no)
        print("Course Name :", self.course_name)


# Driver Code
student = CollegeStudent("Sai", 101, "Computer Science")

student.display_details()
```

## 7. Company Hierarchy (Medium)
### Grand Parent
### Company
### Attribute:
#### • company_name
### Parent
### Department
### Attribute:
#### • department_name
### Child
### Employee
### Attribute:
#### • employee_name
### Display all information.
### Requirement:
#### Use super().

```python
# Grand Parent Class
class Company:
    def __init__(self, company_name):
        self.company_name = company_name


# Parent Class
class Department(Company):
    def __init__(self, company_name, department_name):
        super().__init__(company_name)   # Calls Company constructor
        self.department_name = department_name


# Child Class
class Employee(Department):
    def __init__(self, company_name, department_name, employee_name):
        super().__init__(company_name, department_name)   # Calls Department constructor
        self.employee_name = employee_name

    def display_information(self):
        print("Company Name    :", self.company_name)
        print("Department Name :", self.department_name)
        print("Employee Name   :", self.employee_name)


# Driver Code
emp = Employee(
    "Infosys",
    "Software Development",
    "Sai"
)

emp.display_information()
```

## 8. Banking System (Medium)
### Abstract Class
### Bank
### Attribute:
#### • bank_name
### Abstract Method:
#### • show_interest()
### Parent
### SBI
### Attribute:
#### • branch_name
### Child
### Customer

### Attribute:
#### • customer_name
### Display all details.
### Requirement:
#### Use Abstraction + Multilevel Inheritance + super().

```python
from abc import ABC, abstractmethod

# Abstract Class
class Bank(ABC):
    def __init__(self, bank_name):
        self.bank_name = bank_name

    @abstractmethod
    def show_interest(self):
        pass


# Parent Class
class SBI(Bank):
    def __init__(self, bank_name, branch_name):
        super().__init__(bank_name)     # Calls Bank constructor
        self.branch_name = branch_name

    def show_interest(self):
        print("Interest Rate : 6.5% per annum")


# Child Class
class Customer(SBI):
    def __init__(self, bank_name, branch_name, customer_name):
        super().__init__(bank_name, branch_name)   # Calls SBI constructor
        self.customer_name = customer_name

    def display_details(self):
        print("Bank Name     :", self.bank_name)
        print("Branch Name   :", self.branch_name)
        print("Customer Name :", self.customer_name)
        self.show_interest()


# Driver Code
customer = Customer(
    "State Bank of India",
    "Hyderabad Branch",
    "Sai"
)

customer.display_details()
```

## 9. E-Commerce System (Medium)
### Grand Parent
### Product
### Attribute:
#### • product_name
### Parent
### Electronics
### Attribute:
#### • warranty
### Child
### Mobile
### Attribute:
#### • price
### Display all details.
### Requirement:
#### Without super().

```python
# Grand Parent Class
class Product:
    def __init__(self, product_name):
        self.product_name = product_name


# Parent Class
class Electronics(Product):
    def __init__(self, product_name, warranty):
        # Calling Product constructor directly
        Product.__init__(self, product_name)
        self.warranty = warranty


# Child Class
class Mobile(Electronics):
    def __init__(self, product_name, warranty, price):
        # Calling Electronics constructor directly
        Electronics.__init__(self, product_name, warranty)
        self.price = price

    def display_details(self):
        print("Product Name :", self.product_name)
        print("Warranty    :", self.warranty)
        print("Price       : ₹", self.price)


# Driver Code
mobile = Mobile("Samsung Galaxy S25", "2 Years", 85000)

mobile.display_details()
```

## 10. Hospital Management System (Hard)
### Abstract Class
### Hospital
### Attribute:
#### • hospital_name
### Abstract Method:
#### • show_specialization()
### Parent
### Doctor
### Attribute:
#### • doctor_name
### Child
### Cardiologist
### Attribute:
#### • experience
### Methods:
#### • show_all_details()
### Requirement:
#### • Abstraction
#### • Multilevel Inheritance
#### • super()
#### • Method Overriding

```python
from abc import ABC, abstractmethod

# Abstract Class
class Hospital(ABC):
    def __init__(self, hospital_name):
        self.hospital_name = hospital_name

    @abstractmethod
    def show_specialization(self):
        pass


# Parent Class
class Doctor(Hospital):
    def __init__(self, hospital_name, doctor_name):
        super().__init__(hospital_name)      # Calls Hospital constructor
        self.doctor_name = doctor_name

    # Method Overriding
    def show_specialization(self):
        print("Specialization : General Medicine")


# Child Class
class Cardiologist(Doctor):
    def __init__(self, hospital_name, doctor_name, experience):
        super().__init__(hospital_name, doctor_name)   # Calls Doctor constructor
        self.experience = experience

    # Method Overriding
    def show_specialization(self):
        print("Specialization : Cardiology")

    def show_all_details(self):
        print("Hospital Name  :", self.hospital_name)
        print("Doctor Name    :", self.doctor_name)
        print("Experience     :", self.experience, "Years")
        self.show_specialization()


# Driver Code
doctor = Cardiologist(
    "Apollo Hospital",
    "Dr. Sai",
    12
)

doctor.show_all_details()
```
