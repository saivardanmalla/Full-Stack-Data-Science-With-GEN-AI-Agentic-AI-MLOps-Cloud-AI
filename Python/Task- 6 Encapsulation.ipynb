## 1. Temperature System
### Create a Temperature class.
#### • Private attribute: __temp
#### • Temperature cannot be less than -50 or greater than 60
#### • Display temperature

```python
class Temperature:
    def __init__(self, temp):
        self.__temp = None
        self.set_temperature(temp)

    # Setter method
    def set_temperature(self, temp):
        if -50 <= temp <= 60:
            self.__temp = temp
        else:
            print("Temperature must be between -50 and 60 degrees Celsius")

    # Getter method
    def get_temperature(self):
        return self.__temp

    # Display method
    def display_temperature(self):
        if self.__temp is not None:
            print("Temperature:", self.__temp, "°C")
        else:
            print("Invalid temperature")


# Creating objects
t1 = Temperature(25)
t1.display_temperature()

t2 = Temperature(80)
t2.display_temperature()
```

## 9. Library Book
### Create a Book class.
### Private Attributes:
#### __book_name
#### __copies
### Validation:
#### • Copies cannot be negative
### Methods:
#### • issue_book()
#### • return_book()

```python
class Book:
    def __init__(self, book_name, copies):
        self.__book_name = book_name
        self.__copies = copies if copies >= 0 else 0

    def issue_book(self):
        if self.__copies > 0:
            self.__copies -= 1
            print("Book issued")
        else:
            print("Book not available")

    def return_book(self):
        self.__copies += 1

    def display_book(self):
        print("Book:", self.__book_name)
        print("Copies:", self.__copies)


book = Book("Python Programming", 5)

book.issue_book()
book.return_book()
book.display_book()
```

## 11. Bank Account System
### Create a BankAccount class.
### Private Attributes:
#### __account_holder
#### __balance
### Methods:
#### deposit()
#### withdraw()
#### check_balance()
### Validation:
#### • Deposit > 0
#### • Withdraw > 0
#### • Withdraw <= balance

```python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.__account_holder = account_holder
        self.__balance = balance if balance >= 0 else 0

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposit Successful")
        else:
            print("Deposit amount must be positive")

    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawal Successful")
        else:
            print("Invalid withdrawal")

    def check_balance(self):
        print("Account Holder:", self.__account_holder)
        print("Balance:", self.__balance)


account = BankAccount("Sai", 10000)

account.deposit(5000)
account.withdraw(3000)
account.check_balance()
```

## 12. Student Result Management
### Create a Student class.
### Private Attributes:
#### __name
#### __python_marks
#### __sql_marks
#### __statistics_marks
### Methods:
#### update_marks()
#### calculate_total()
#### calculate_percentage()
#### display_result()
### Validation:
#### • Marks between 0 and 100

```python
class Student:
    def __init__(self, name, python, sql, statistics):
        self.__name = name
        self.__python_marks = 0
        self.__sql_marks = 0
        self.__statistics_marks = 0

        self.update_marks(python, sql, statistics)

    def update_marks(self, python, sql, statistics):
        if 0 <= python <= 100 and 0 <= sql <= 100 and 0 <= statistics <= 100:
            self.__python_marks = python
            self.__sql_marks = sql
            self.__statistics_marks = statistics
        else:
            print("Marks should be between 0 and 100")

    def calculate_total(self):
        return self.__python_marks + self.__sql_marks + self.__statistics_marks

    def calculate_percentage(self):
        return self.calculate_total() / 3

    def display_result(self):
        print("Student Name:", self.__name)
        print("Total Marks:", self.calculate_total())
        print("Percentage:", self.calculate_percentage(), "%")


student = Student("Sai", 90, 85, 95)

student.display_result()
```

## 13. E-commerce Product
### Create a Product class.
### Private Attributes:
#### __product_name
#### __price
#### __stock
### Methods:
#### update_stock()
#### purchase()
#### display_product()
### Validation:
#### • Price > 0
#### • Stock >= 0

```python
class Product:
    def __init__(self, product_name, price, stock):
        self.__product_name = product_name
        self.__price = price if price > 0 else 0
        self.__stock = stock if stock >= 0 else 0

    def update_stock(self, quantity):
        if quantity >= 0:
            self.__stock += quantity

    def purchase(self, quantity):
        if quantity <= self.__stock:
            self.__stock -= quantity
            print("Purchase Successful")
        else:
            print("Not enough stock")

    def display_product(self):
        print("Product:", self.__product_name)
        print("Price:", self.__price)
        print("Stock:", self.__stock)


product = Product("Laptop", 50000, 10)

product.purchase(2)
product.display_product()
```

## 16. ATM Machine
### Features:
#### • Private PIN
#### • Private Balance
### Methods:
#### deposit()
#### withdraw()
#### change_pin()
#### check_balance()
### Validation:
#### • Correct PIN required
#### • Withdraw <= Balance

```python
class ATM:
    def __init__(self, pin, balance):
        self.__pin = pin
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposit Successful")
        else:
            print("Invalid Deposit Amount")

    def withdraw(self, pin, amount):
        if pin == self.__pin:
            if amount > 0 and amount <= self.__balance:
                self.__balance -= amount
                print("Withdrawal Successful")
            else:
                print("Insufficient Balance")
        else:
            print("Incorrect PIN")

    def change_pin(self, old_pin, new_pin):
        if old_pin == self.__pin:
            self.__pin = new_pin
            print("PIN Changed Successfully")
        else:
            print("Wrong PIN")

    def check_balance(self, pin):
        if pin == self.__pin:
            print("Balance:", self.__balance)
        else:
            print("Incorrect PIN")


atm = ATM(1234, 10000)

atm.deposit(5000)
atm.withdraw(1234, 3000)
atm.check_balance(1234)
atm.change_pin(1234, 5678)
```

## 18. Laptop Store
### Private Attributes:
#### __brand
#### __ram
#### __price
#### __stock
### Validation:
#### • RAM in [4, 8, 16, 32, 64]
#### • Price > 0
### Methods:
#### sell_laptop()
#### restock()
#### display_details()

```python
class Laptop:
    def __init__(self, brand, ram, price, stock):
        self.__brand = brand

        if ram in [4, 8, 16, 32, 64]:
            self.__ram = ram
        else:
            self.__ram = 4

        self.__price = price if price > 0 else 0
        self.__stock = stock if stock >= 0 else 0

    def sell_laptop(self):
        if self.__stock > 0:
            self.__stock -= 1
            print("Laptop Sold")
        else:
            print("Out of Stock")

    def restock(self, quantity):
        self.__stock += quantity

    def display_details(self):
        print("Brand:", self.__brand)
        print("RAM:", self.__ram, "GB")
        print("Price:", self.__price)
        print("Stock:", self.__stock)


laptop = Laptop("HP", 16, 60000, 5)

laptop.sell_laptop()
laptop.display_details()
```

## 20. Mini Banking Project
### Create a complete banking system.
### Private Attributes:
#### __account_number
#### __account_holder
#### __balance
### Methods:
#### deposit()
#### withdraw()
#### transfer()
#### check_balance()
### Validation:
#### • Deposit > 0
#### • Withdraw > 0
#### • Transfer amount <= balance

```python
class BankAccount:

    def __init__(self, account_number, account_holder, balance):
        self.__account_number = account_number
        self.__account_holder = account_holder
        self.__balance = balance if balance >= 0 else 0


    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposit Successful")
        else:
            print("Invalid Deposit")


    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawal Successful")
        else:
            print("Invalid Withdrawal")


    def transfer(self, receiver, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            receiver.__balance += amount
            print("Transfer Successful")
        else:
            print("Transfer Failed")


    def check_balance(self):
        print("Account Holder:", self.__account_holder)
        print("Balance:", self.__balance)



# Creating Accounts
account1 = BankAccount(1001, "Sai", 50000)
account2 = BankAccount(1002, "Rahul", 20000)


account1.deposit(10000)
account1.withdraw(5000)

account1.transfer(account2, 15000)

account1.check_balance()
account2.check_balance()
```
