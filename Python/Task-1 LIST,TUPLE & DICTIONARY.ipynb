## Dictionary Questions

### 1. Create a dictionary with student name and marks

```python
student = {
    "Sai": 90,
    "Srusti": 85,
    "Srujan": 95
}

print(student)
```

### 2. Access a value using key

```python
student = {"Sai": 90, "Amit": 80}

print(student["Sai"])
```

### 3. Add a new key-value pair into dictionary


```python
student = {"Sai": 90}

student["Rahul"] = 85

print(student)
```

### 4. Update an existing value in dictionary

```python
student = {"Sai": 90}

student["Sai"] = 95

print(student)
```

### 5. Remove a key using pop()

```python
student = {"Sai":90, "Amit":80}

student.pop("Amit")

print(student)
```

### 6. Remove the last inserted item using popitem()

```python
student = {"Sai":90, "Amit":80}

student.popitem()

print(student)
```

### 7. Display all keys from dictionary

```python
student = {"Sai":90, "Amit":80}

print(student.keys())
```

### 8. Display all values from dictionary

```python
student = {"Sai":90, "Amit":80}

print(student.values())
```

### 9. Display all items from dictionary


```python
student = {"Sai":90, "Amit":80}

print(student.items())
```

### 10. Check whether a key exists in dictionary

```python
student = {"Sai":90, "Amit":80}

if "Sai" in student:
    print("Key exists")
else:
    print("Key not found")
```

### 11. Find length of dictionary

```python
student = {"Sai":90, "Amit":80, "Rahul":70}

print(len(student))
```

### 12. Copy one dictionary into another

```python
student1 = {"Sai":90}

student2 = student1.copy()

print(student2)
```

### 13. Clear all items from dictionary

```python
student = {"Sai":90, "Amit":80}

student.clear()

print(student)
```

### 14. Create dictionary using fromkeys()

```python
keys = ["Name", "Age", "Branch"]

student = dict.fromkeys(keys)

print(student)
```

### 15. Get value of a key using get()

```python
student = {"Sai":90}

print(student.get("Sai"))
```

### 16. Iterate through dictionary using loop

```python
student = {
    "Sai":90,
    "Amit":80
}

for key, value in student.items():
    print(key, value)
```

### 17. Merge two dictionaries


```python
dict1 = {"A":10, "B":20}
dict2 = {"C":30, "D":40}

dict1.update(dict2)

print(dict1)
```

### 18. Create nested dictionary and access inner values

```python
student = {
    "student1":{
        "name":"Sai",
        "marks":90
    }
}

print(student["student1"]["marks"])
```

### 19. Convert dictionary keys into list

```python
student = {
    "Sai":90,
    "Amit":80,
    "Rahul":70
}

keys_list = list(student.keys())

print(keys_list)
```

### 20. Sort dictionary keys and display them

```python
student = {
    "Srujan":90,
    "Amit":80,
    "Sai":95
}

sorted_keys = sorted(student.keys())

print(sorted_keys)
```
