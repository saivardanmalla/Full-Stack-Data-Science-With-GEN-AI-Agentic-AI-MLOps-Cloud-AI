## 1. Vehicle Start Method
### Question:
#### Create an abstract class Vehicle with an abstract method start().
#### Create a child class Car and implement the start() method.
### Concept:
#### The parent class defines what should happen, but the child class decides how.

```python
from abc import ABC, abstractmethod

# Abstract class
class Vehicle(ABC):

    # Abstract method
    @abstractmethod
    def start(self):
        pass


# Child class
class Car(Vehicle):

    # Implementing abstract method
    def start(self):
        print("Car starts using a key or push button")


# Main program

car = Car()

# Calling method
car.start()
```

## 2. Animal Sound
### Question:
#### Create an abstract class Animal with an abstract method sound().
#### Create child classes:
##### • Dog
##### • Cat
#### Implement the sound() method in both classes.
### Concept:
#### Different classes provide different implementations of the same method.

```python
from abc import ABC, abstractmethod

# Abstract class
class Animal(ABC):

    # Abstract method
    @abstractmethod
    def sound(self):
        pass


# Child class Dog
class Dog(Animal):

    def sound(self):
        print("Dog says: Bark")


# Child class Cat
class Cat(Animal):

    def sound(self):
        print("Cat says: Meow")


# Main program
dog = Dog()
cat = Cat()

# Calling methods
dog.sound()
cat.sound()
```

## 3. Shape Area Calculation
### Question:
#### Create an abstract class Shape with an abstract method area().
### Create:
#### • Rectangle
#### • Circle
### Implement the area() method.
### Concept:
#### Every shape calculates area differently.

```python

from abc import ABC, abstractmethod
import math

# Abstract class
class Shape(ABC):

    # Abstract method
    @abstractmethod
    def area(self):
        pass


# Rectangle class
class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    # Implementing area method
    def area(self):
        return self.length * self.width


# Circle class
class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    # Implementing area method
    def area(self):
        return math.pi * self.radius * self.radius


# Main program
rectangle = Rectangle(10, 5)
circle = Circle(7)

print("Rectangle Area:", rectangle.area())
print("Circle Area:", circle.area())
```

## 6. Online Food Delivery App
### Question:

#### • order_food()
#### • track_order()

## 7. Bank Transaction System
### Question:
### Create an abstract class BankTransaction.
### Abstract methods:
#### • deposit()
#### • withdraw()
### Create:
#### • SBI
#### • HDFC
### Implement both methods.
### Concept:
#### Different banks have different transaction rules.

```python
from abc import ABC, abstractmethod

# Abstract class
class BankTransaction(ABC):

    @abstractmethod
    def deposit(self):
        pass

    @abstractmethod
    def withdraw(self):
        pass


# SBI class
class SBI(BankTransaction):

    def deposit(self):
        print("Deposit money in SBI account")

    def withdraw(self):
        print("Withdraw money from SBI account with SBI rules")


# HDFC class
class HDFC(BankTransaction):

    def deposit(self):
        print("Deposit money in HDFC account")

    def withdraw(self):
        print("Withdraw money from HDFC account with HDFC rules")


# Main program
sbi = SBI()
hdfc = HDFC()

sbi.deposit()
sbi.withdraw()

hdfc.deposit()
hdfc.withdraw()
```

## 8. Notification System
### Question:
### Create an abstract class Notification.
### Abstract method:
#### • send_notification()
### Create:
#### • EmailNotification
#### • SMSNotification
#### • WhatsAppNotification
### Implement the method in all classes.
### Concept:
#### One common interface, multiple implementations.


```python
from abc import ABC, abstractmethod

# Abstract class
class Notification(ABC):

    @abstractmethod
    def send_notification(self):
        pass


# Email Notification class
class EmailNotification(Notification):

    def send_notification(self):
        print("Sending notification through Email")


# SMS Notification class
class SMSNotification(Notification):

    def send_notification(self):
        print("Sending notification through SMS")


# WhatsApp Notification class
class WhatsAppNotification(Notification):

    def send_notification(self):
        print("Sending notification through WhatsApp")


# Main program
email = EmailNotification()
sms = SMSNotification()
whatsapp = WhatsAppNotification()

email.send_notification()
sms.send_notification()
whatsapp.send_notification()
```


## 9. E-Commerce Shipping
### Question:
### Create an abstract class Shipping.
### Abstract method:
#### • calculate_charges()
### Create:
#### • StandardShipping
#### • ExpressShipping
### Implement different charge calculations.
### Concept:
#### Different shipping methods calculate charges differently.


```python
from abc import ABC, abstractmethod

# Abstract class
class Shipping(ABC):

    @abstractmethod
    def calculate_charges(self):
        pass


# Standard Shipping class
class StandardShipping(Shipping):

    def calculate_charges(self):
        print("Standard Shipping Charge: ₹50")


# Express Shipping class
class ExpressShipping(Shipping):

    def calculate_charges(self):
        print("Express Shipping Charge: ₹150")


# Main program
standard = StandardShipping()
express = ExpressShipping()

standard.calculate_charges()
express.calculate_charges()
```


## 10. AI Model Prediction System
### Question:
### Create an abstract class Model.
### Abstract method:
#### • predict()
### Create:
#### • RecommendationModel
#### • FraudDetectionModel
### Implement the predict() method.
### Concept:
#### Different AI models make predictions in different ways.

```python
from abc import ABC, abstractmethod

# Abstract class
class Model(ABC):

    @abstractmethod
    def predict(self):
        pass


# Recommendation Model
class RecommendationModel(Model):

    def predict(self):
        print("Recommendation Model suggests products based on user interests")


# Fraud Detection Model
class FraudDetectionModel(Model):

    def predict(self):
        print("Fraud Detection Model identifies suspicious transactions")


# Main program
recommendation = RecommendationModel()
fraud_detection = FraudDetectionModel()

recommendation.predict()
fraud_detection.predict()
```
