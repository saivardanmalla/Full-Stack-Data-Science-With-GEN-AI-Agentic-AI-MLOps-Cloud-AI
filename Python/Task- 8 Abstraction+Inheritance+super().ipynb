# Abstraction + Inheritance + super()

## 1. Vehicle System (Easy)
### Question
### Create an abstract class Vehicle.
### Parent Class:
#### • brand
### Abstract Method:
#### • start()
### Child Class:
#### • Car
### Requirements:
#### • Use super() to initialize brand.
#### • Implement start().
### Concept
#### Learn how super() calls the parent constructor and how abstraction forces the child
### class to implement methods.

```python
from abc import ABC, abstractmethod

# Abstract Parent Class
class Vehicle(ABC):
    def __init__(self, brand):
        self.brand = brand

    @abstractmethod
    def start(self):
        pass


# Child Class
class Car(Vehicle):
    def __init__(self, brand):
        super().__init__(brand)   # Calling parent constructor

    def start(self):
        print(f"{self.brand} car is starting...")


# Object Creation
c1 = Car("Toyota")

print("Brand:", c1.brand)
c1.start()
```

## 2. Employee Management (Easy)
### Question
### Create an abstract class Employee.
### Attributes:
#### • name
#### • salary
### Abstract Method:
#### • work()
### Child Class:
#### • Developer
### Requirements:
#### • Use super() for name and salary.
#### • Implement work().
### Concept
#### Pass common data to the parent class using super().

```python
from abc import ABC, abstractmethod

# Abstract Class
class Employee(ABC):
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @abstractmethod
    def work(self):
        pass


# Child Class
class Developer(Employee):
    def __init__(self, name, salary):
        super().__init__(name, salary)   # Calls parent constructor

    def work(self):
        print(f"{self.name} is developing software.")
        print(f"Salary: ₹{self.salary}")


# Driver Code
emp = Developer("Sai", 50000)
emp.work()
```

## 3. Bank Account (Easy)
### Question
### Create an abstract class Bank.
### Attributes:
#### • account_holder
#### • account_number
### Abstract Method:
#### • show_interest_rate()
### Child Class:
#### • SBI
### Requirements:
#### • Use super().
#### • Print SBI interest rate.
### Concept
#### Combine abstraction and constructor inheritance.

```python
from abc import ABC, abstractmethod

# Abstract Class
class Bank(ABC):
    def __init__(self, account_holder, account_number):
        self.account_holder = account_holder
        self.account_number = account_number

    @abstractmethod
    def show_interest_rate(self):
        pass


# Child Class
class SBI(Bank):
    def __init__(self, account_holder, account_number):
        super().__init__(account_holder, account_number)   # Calls parent constructor

    def show_interest_rate(self):
        print("Bank Name       : SBI")
        print("Account Holder  :", self.account_holder)
        print("Account Number  :", self.account_number)
        print("Interest Rate   : 6.5% per annum")


# Driver Code
customer = SBI("Sai", "1234567890")
customer.show_interest_rate()
```

## 5. Shape Area Calculator (Easy)
### Question
### Create an abstract class Shape.
### Attribute:
#### • color
### Abstract Method:
#### • area()
#### Child Class:
#### • Rectangle
### Requirements:
#### • Use super().
#### • Calculate rectangle area.
### Concept
#### Use parent constructor and child-specific logic.

```python
from abc import ABC, abstractmethod

# Abstract Class
class Shape(ABC):
    def __init__(self, color):
        self.color = color

    @abstractmethod
    def area(self):
        pass


# Child Class
class Rectangle(Shape):
    def __init__(self, color, length, width):
        super().__init__(color)      # Calls parent constructor
        self.length = length
        self.width = width

    def area(self):
        area = self.length * self.width
        print("Shape      : Rectangle")
        print("Color      :", self.color)
        print("Length     :", self.length)
        print("Width      :", self.width)
        print("Area       :", area)


# Driver Code
rect = Rectangle("Blue", 10, 5)
rect.area()
```

## 6. Online Payment System
### Question
### Create an abstract class Payment.
### Attributes:
#### • user_name
#### • amount
### Abstract Method:
#### • make_payment()
### Child Classes:
#### • UPI
#### • CreditCard
### Requirements:
#### • Use super().
#### • Implement payment logic.
### Concept
#### Multiple child classes implementing the same abstract method.

```python
from abc import ABC, abstractmethod

# Abstract Class
class Payment(ABC):
    def __init__(self, user_name, amount):
        self.user_name = user_name
        self.amount = amount

    @abstractmethod
    def make_payment(self):
        pass


# Child Class 1
class UPI(Payment):
    def __init__(self, user_name, amount):
        super().__init__(user_name, amount)

    def make_payment(self):
        print("----- UPI Payment -----")
        print("User Name :", self.user_name)
        print("Amount    : ₹", self.amount)
        print("Payment Successful using UPI")


# Child Class 2
class CreditCard(Payment):
    def __init__(self, user_name, amount):
        super().__init__(user_name, amount)

    def make_payment(self):
        print("----- Credit Card Payment -----")
        print("User Name :", self.user_name)
        print("Amount    : ₹", self.amount)
        print("Payment Successful using Credit Card")


# Driver Code
payment1 = UPI("Sai", 1500)
payment2 = CreditCard("Rahul", 2500)

payment1.make_payment()
print()
payment2.make_payment()
```

## 7. School Management System
### Question
### Create an abstract class School.
### Attributes:
#### • school_name
#### • city
### Abstract Method:
#### • show_department()
### Child Class:
#### • EngineeringCollege
### Requirements:
#### • Use super().
#### • Display department details.
### Concept
#### Constructor chaining with abstraction.

```python
from abc import ABC, abstractmethod

# Abstract Class
class School(ABC):
    def __init__(self, school_name, city):
        self.school_name = school_name
        self.city = city

    @abstractmethod
    def show_department(self):
        pass


# Child Class
class EngineeringCollege(School):
    def __init__(self, school_name, city, department):
        super().__init__(school_name, city)   # Calls parent constructor
        self.department = department

    def show_department(self):
        print("School Name :", self.school_name)
        print("City        :", self.city)
        print("Department  :", self.department)


# Driver Code
college = EngineeringCollege(
    "ABC Engineering College",
    "Hyderabad",
    "Computer Science and Engineering"
)

college.show_department()
```

## 8. Food Delivery App
### Question
### Create an abstract class FoodDelivery.
### Attributes:
#### • customer_name
#### • order_id
### Abstract Methods:
#### • place_order()
#### • track_order()
### Child Class:
#### • Swiggy
### Requirements:
#### • Use super().
#### • Implement both methods.
### Concept
#### Multiple abstract methods in one parent class.

```python
from abc import ABC, abstractmethod

# Abstract Class
class FoodDelivery(ABC):
    def __init__(self, customer_name, order_id):
        self.customer_name = customer_name
        self.order_id = order_id

    @abstractmethod
    def place_order(self):
        pass

    @abstractmethod
    def track_order(self):
        pass


# Child Class
class Swiggy(FoodDelivery):
    def __init__(self, customer_name, order_id):
        super().__init__(customer_name, order_id)   # Calls parent constructor

    def place_order(self):
        print("Customer Name :", self.customer_name)
        print("Order ID      :", self.order_id)
        print("Order placed successfully on Swiggy.")

    def track_order(self):
        print("Tracking Order...")
        print("Order Status  : Your order is on the way.")


# Driver Code
order = Swiggy("Sai", 1001)

order.place_order()
print()
order.track_order()
```

## 9. AI Model Prediction System
### Question
### Create an abstract class Model.
### Attributes:
#### • model_name
#### • version
### Abstract Method:
#### • predict()
### Child Class:
#### • RecommendationModel
### Requirements:
### • Use super().
### • Implement prediction method.
#### Concept
### Real-world AI abstraction example.

```python
from abc import ABC, abstractmethod

# Abstract Class
class Model(ABC):
    def __init__(self, model_name, version):
        self.model_name = model_name
        self.version = version

    @abstractmethod
    def predict(self):
        pass


# Child Class
class RecommendationModel(Model):
    def __init__(self, model_name, version):
        super().__init__(model_name, version)   # Calls parent constructor

    def predict(self):
        print("Model Name :", self.model_name)
        print("Version    :", self.version)
        print("Prediction : Recommended Product -> Wireless Headphones")


# Driver Code
model = RecommendationModel("Product Recommendation Model", "v1.0")
model.predict()
```

## 10. Hospital Management System
### Question
### Create an abstract class Hospital.
### Attributes:
#### • hospital_name
#### • location
### Abstract Method:
#### • show_specialization()
### Child Class:
#### • CardiologyDepartment
### Requirements:
### • Use super().
### • Display specialization details.
### Concept
#### Practice abstraction with real-world hierarchy.

```python
from abc import ABC, abstractmethod

# Abstract Class
class Hospital(ABC):
    def __init__(self, hospital_name, location):
        self.hospital_name = hospital_name
        self.location = location

    @abstractmethod
    def show_specialization(self):
        pass


# Child Class
class CardiologyDepartment(Hospital):
    def __init__(self, hospital_name, location, specialization):
        super().__init__(hospital_name, location)   # Calls parent constructor
        self.specialization = specialization

    def show_specialization(self):
        print("Hospital Name :", self.hospital_name)
        print("Location      :", self.location)
        print("Department    : Cardiology")
        print("Specialization:", self.specialization)


# Driver Code
hospital = CardiologyDepartment(
    "Apollo Hospital",
    "Hyderabad",
    "Heart Disease Diagnosis and Treatment"
)

hospital.show_specialization()
```
