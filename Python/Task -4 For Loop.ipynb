## 1. Print Names Starting With "A"
### names = ["Aman", "Ravi", "Ankit", "Sumit", "Ajay"]
### Print only names starting with "A".

```python
names = ["Aman", "Ravi", "Ankit", "Sumit", "Ajay"]

for name in names:
    if name.startswith("A"):
        print(name)
```

## 3. Count Names Containing "a"
### names = ["Rahul", "Aman", "Riya", "Karan", "Sonia"]
### Count how many names contain letter "a".

```python
names = ["Rahul", "Aman", "Riya", "Karan", "Sonia"]

count = 0

for name in names:
    if "a" in name.lower():
        count += 1

print("Count:", count)
```

## 11. Count Even and Odd Numbers
### nums = [11, 22, 33, 44, 55, 66]
#### Count how many even and odd numbers are present.

```python
nums = [11, 22, 33, 44, 55, 66]

even = 0
odd = 0

for num in nums:
    if num % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even:", even)
print("Odd:", odd)
```

## 13. Print Palindrome Numbers
### nums = [121, 234, 1331, 456, 787]
#### Print numbers whose reverse is same.

```python
nums = [121, 234, 1331, 456, 787]

for num in nums:

    original = num
    reverse = 0

    while num > 0:
        digit = num % 10
        reverse = reverse * 10 + digit
        num //= 10

    if original == reverse:
        print(original)
```

## 14. Find Longest Word
### words = ["data", "science", "python", "artificialintelligence"]
#### Find the longest word using a loop.

```python
words = ["data", "science", "python", "artificialintelligence"]

longest = words[0]

for word in words:
    if len(word) > len(longest):
        longest = word

print("Longest Word:", longest)
```

## 19. Multiplication Table for Each Number
### nums = [2, 3, 5]
### Print multiplication table from 1 to 5 for every number.

```python
nums = [2, 3, 5]

for num in nums:

    print("Table of", num)

    for i in range(1, 6):
        print(num, "x", i, "=", num*i)

    print()
```

## 20. Find Second Largest Number
### nums = [12, 45, 67, 89, 34, 89, 76]
### Find the second largest number using loops only.

```python
nums = [12, 45, 67, 89, 34, 89, 76]

largest = second = float('-inf')

for num in nums:

    if num > largest:
        second = largest
        largest = num

    elif num > second and num != largest:
        second = num

print("Second Largest:", second)
```
