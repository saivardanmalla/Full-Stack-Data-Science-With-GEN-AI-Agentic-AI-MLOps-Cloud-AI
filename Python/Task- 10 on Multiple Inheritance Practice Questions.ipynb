# Multiple Inheritance

## 1. Father + Mother → Child (Easy)
### Father
#### • father_name
#### • father_age
### Mother
#### • mother_name
#### • mother_age
### Child
#### • child_name
### Display all details.
#### Requirement: Call parent constructors directly.

```python
# Parent Class 1
class Father:
    def __init__(self, father_name, father_age):
        self.father_name = father_name
        self.father_age = father_age


# Parent Class 2
class Mother:
    def __init__(self, mother_name, mother_age):
        self.mother_name = mother_name
        self.mother_age = mother_age


# Child Class
class Child(Father, Mother):
    def __init__(self, father_name, father_age,
                 mother_name, mother_age,
                 child_name):

        # Calling parent constructors directly
        Father.__init__(self, father_name, father_age)
        Mother.__init__(self, mother_name, mother_age)

        self.child_name = child_name

    def display_details(self):
        print("Father Name :", self.father_name)
        print("Father Age  :", self.father_age)
        print("Mother Name :", self.mother_name)
        print("Mother Age  :", self.mother_age)
        print("Child Name  :", self.child_name)


# Driver Code
child = Child(
    "Ramesh", 48,
    "Sita", 45,
    "Sai"
)

child.display_details()
```

## 2. Teacher + Engineer → Student
### Teacher
#### • teaching_subject
### Engineer
#### • programming_language
### Student
#### • student_name
### Display all details.
#### Requirement: Use parent class names.

```python
# Teacher + Engineer -> Student

class Teacher:

    def __init__(self, teacher_subject, **kwargs):
        self.teacher_subject = teacher_subject
        super().__init__(**kwargs)


class Engineer:

    def __init__(self, programming_language, **kwargs):
        self.programming_language = programming_language
        super().__init__(**kwargs)



class Student(Teacher, Engineer):

    def __init__(self, teacher_subject, programming_language, student_name):

        super().__init__(
            teacher_subject=teacher_subject,
            programming_language=programming_language
        )

        self.student_name = student_name


    def show(self):

        print("Teacher Subject:", self.teacher_subject)
        print("Programming Language:", self.programming_language)
        print("Student Name:", self.student_name)



e = Student(
    teacher_subject="Database Management System",
    programming_language="Python Programming Language",
    student_name="Malla Sai Vardan"
)


e.show()
```

# Multiple Inheritance + Abstraction
## 11. Bank + Insurance → Customer
### Abstract Class
### Bank
### Abstract Method:
#### • show_bank_details()
### Attribute:
#### • bank_name
### Insurance
### Attribute:
#### • policy_name
#### Customer
### Attribute:
#### • customer_name
### Implement abstract method and display all details.

```python
from abc import ABC, abstractmethod

# Abstract Parent Class
class Bank(ABC):
    def __init__(self, bank_name):
        self.bank_name = bank_name

    @abstractmethod
    def show_bank_details(self):
        pass


# Parent Class
class Insurance:
    def __init__(self, policy_name):
        self.policy_name = policy_name


# Child Class
class Customer(Bank, Insurance):
    def __init__(self, bank_name, policy_name, customer_name):
        # Calling parent constructors directly
        Bank.__init__(self, bank_name)
        Insurance.__init__(self, policy_name)

        self.customer_name = customer_name

    # Implementing abstract method
    def show_bank_details(self):
        print("Bank Name     :", self.bank_name)

    def display_details(self):
        self.show_bank_details()
        print("Policy Name   :", self.policy_name)
        print("Customer Name :", self.customer_name)


# Driver Code
customer = Customer(
    "State Bank of India",
    "Life Insurance Policy",
    "Sai"
)

customer.display_details()
```

## 12. Animal + Pet → Dog
### Abstract Class
### Animal
### Attribute:
#### • animal_name
#### Abstract Method:
#### • make_sound()
### Pet
### Attribute:
#### • owner_name
### Dog
#### Display all information.

```python
from abc import ABC, abstractmethod

# Abstract Parent Class
class Animal(ABC):
    def __init__(self, animal_name):
        self.animal_name = animal_name

    @abstractmethod
    def make_sound(self):
        pass


# Parent Class
class Pet:
    def __init__(self, owner_name):
        self.owner_name = owner_name


# Child Class
class Dog(Animal, Pet):
    def __init__(self, animal_name, owner_name):
        # Calling parent constructors directly
        Animal.__init__(self, animal_name)
        Pet.__init__(self, owner_name)

    # Implementing abstract method
    def make_sound(self):
        print("Sound         : Woof! Woof!")

    def display_details(self):
        print("Animal Name   :", self.animal_name)
        print("Owner Name    :", self.owner_name)
        self.make_sound()


# Driver Code
dog = Dog("German Shepherd", "Sai")

dog.display_details()
```

## 13. Employee + Department → Manager
### Abstract Class
### Employee
### Attribute:
#### • employee_name
### Abstract Method:
#### • work()
### Department
### Attribute:
#### • department_name
### Manager
#### Implement work() and display all details.

```python
from abc import ABC, abstractmethod

# Abstract Parent Class
class Employee(ABC):
    def __init__(self, employee_name):
        self.employee_name = employee_name

    @abstractmethod
    def work(self):
        pass


# Parent Class
class Department:
    def __init__(self, department_name):
        self.department_name = department_name


# Child Class
class Manager(Employee, Department):
    def __init__(self, employee_name, department_name):
        # Calling parent constructors directly
        Employee.__init__(self, employee_name)
        Department.__init__(self, department_name)

    # Implementing abstract method
    def work(self):
        print("Work            : Managing the department and employees.")

    # Display all details
    def display_details(self):
        print("Employee Name   :", self.employee_name)
        print("Department Name :", self.department_name)
        self.work()


# Driver Code
manager = Manager("Sai", "Data Science")

manager.display_details()
```

## 14. Course + Institute → Student## 
### Abstract Class
### Course
### Attribute:
#### • course_name
### Abstract Method:
#### • duration()
### Institute
### Attribute:
#### • institute_name
### Student
### Attribute:
#### • student_name
### Display all details.

```python
from abc import ABC, abstractmethod

# Abstract Parent Class
class Course(ABC):
    def __init__(self, course_name):
        self.course_name = course_name

    @abstractmethod
    def duration(self):
        pass


# Parent Class
class Institute:
    def __init__(self, institute_name):
        self.institute_name = institute_name


# Child Class
class Student(Course, Institute):
    def __init__(self, course_name, institute_name, student_name):
        # Calling parent constructors directly
        Course.__init__(self, course_name)
        Institute.__init__(self, institute_name)

        self.student_name = student_name

    # Implementing abstract method
    def duration(self):
        print("Course Duration : 6 Months")

    # Display all details
    def display_details(self):
        print("Student Name    :", self.student_name)
        print("Course Name     :", self.course_name)
        print("Institute Name  :", self.institute_name)
        self.duration()


# Driver Code
student = Student(
    "Python Full Stack",
    "ABC Institute",
    "Sai"
)

student.display_details()
```
