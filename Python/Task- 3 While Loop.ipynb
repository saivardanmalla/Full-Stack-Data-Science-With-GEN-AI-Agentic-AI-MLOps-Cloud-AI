## 1. Print Numbers from 1 to 20
### Question
### Write a program to print numbers from 1 to 20 using a while loop.
### Explanation
#### • Start from 1

## 4. Reverse a Number
### Question
### Write a program to reverse a number using a while loop.
### Explanation
#### Example:
#### 1234 → 4321
#### Steps:
#### • Take last digit using % 10
#### • Add digit into reverse number
#### • Remove last digit using // 10

```python
num = int(input("Enter number: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reverse =", reverse)
```

## 6. Palindrome Number
### Question
### Write a program to check whether a number is palindrome or not.
### Explanation
#### Palindrome means same from both sides.
#### Example:
#### 121 → Palindrome
#### 1331 → Palindrome
#### Steps:
#### • Reverse the number
#### • Compare with original number

```python
num = int(input("Enter number: "))

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print("Palindrome Number")
else:
    print("Not Palindrome")
```

## 7. Armstrong Number
### Question
### Write a program to check whether a number is Armstrong number.
### Explanation
#### Example:
#### 153
#### Calculation:
#### 13 + 53 + 33 = 153
#### If sum equals original number → Armstrong number

```python
num = int(input("Enter number: "))

original = num
sum = 0

while num > 0:
    digit = num % 10
    sum += digit ** 3
    num = num // 10

if sum == original:
    print("Armstrong Number")
else:
    print("Not Armstrong")
```

## 11. Fibonacci Series
### Question
### Write a program to print Fibonacci series.
#### Explanation
#### Series:
#### 0 1 1 2 3 5 8 13 ...
#### Each next number is sum of previous two numbers.

```python
n = int(input("Enter number of terms: "))

a = 0
b = 1
count = 0

while count < n:
    print(a)
    
    c = a + b
    a = b
    b = c
    
    count += 1
```

## 12. Factorial of Number

### Write a program to find factorial of a number.
#### Explanation
#### Example:
#### 5! = 5 × 4 × 3 × 2 × 1 = 120
#### • Multiply numbers one by one using loop

```python
num = int(input("Enter number: "))

fact = 1

while num > 0:
    fact *= num
    num -= 1

print("Factorial =", fact)
```

## 13. Menu Driven Calculator
### Question
### Create calculator using while loop and if elif else.
### Explanation
#### Menu Example:
##### 1. Add
##### 2. Subtract
##### 3. Multiply
##### 4. Divide
##### 5. Exit
#### • User selects option
#### • Perform operation
#### • Continue until exit option chosen

```python
while True:

    print("1.Add")
    print("2.Subtract")
    print("3.Multiply")
    print("4.Divide")
    print("5.Exit")

    choice = int(input("Enter choice: "))

    if choice == 5:
        break

    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    if choice == 1:
        print(a+b)

    elif choice == 2:
        print(a-b)

    elif choice == 3:
        print(a*b)

    elif choice == 4:
        print(a/b)

    else:
        print("Invalid Choice")
```

## 14. ATM System
### Question
### Create simple ATM program.
### Explanation
### Options:
#### • Check balance
#### • Deposit money
#### • Withdraw money"
#### • Exit
##### Use loop so menu repeats again and again.

```python
balance = 5000

while True:

    print("\n1.Check Balance")
    print("2.Deposit")
    print("3.Withdraw")
    print("4.Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        print("Balance:", balance)

    elif choice == 2:
        amount = int(input("Deposit amount: "))
        balance += amount
        print("Deposited")

    elif choice == 3:
        amount = int(input("Withdraw amount: "))

        if amount <= balance:
            balance -= amount
            print("Withdraw Successful")
        else:
            print("Insufficient Balance")

    elif choice == 4:
        break
```

## 16. Leap Year Checker Repeatedly
### Question
### Check leap year repeatedly until user types "stop".
### Explanation
#### • Ask year from user
#### • Check leap year condition
#### • Continue until "stop" entered
### Leap year condition:
#### • Divisible by 4
#### • Not divisible by 100 OR divisible by 400

```python
while True:

    year = input("Enter year or stop: ")

    if year == "stop":
        break

    year = int(year)

    if year % 4 == 0 and (year % 100 != 0 or year % 400 == 0):
        print("Leap Year")
    else:
        print("Not Leap Year")
```

## 18. Prime Numbers Between 1 to 100
### Question
### Print all prime numbers from 1 to 100.
### Explanation
#### Prime number:
#### • Divisible only by 1 and itself
### Example:
#### 2, 3, 5, 7, 11 ...
### Use nested loop to check divisibility.

```python
num = 2

while num <= 100:

    count = 0
    i = 1

    while i <= num:
        if num % i == 0:
            count += 1
        i += 1

    if count == 2:
        print(num)

    num += 1
```

## 19. Student Marks Average
### Question
### Keep taking student marks until negative number entered.
### Explanation
#### • Add all marks
#### • Count total students
#### • Stop on negative input
#### • Find average:
### average = total / count

```python
total = 0
count = 0

while True:

    marks = int(input("Enter marks: "))

    if marks < 0:
        break

    total += marks
    count += 1


average = total / count

print("Average Marks:", average)
```

## 20. Login System with 3 Attempts
### Question
### Create login system with maximum 3 attempts.
### Explanation
#### • Store correct username and password
#### • Ask user input
#### • If correct → Login successful
#### • If wrong → decrease attempts
#### • After 3 wrong attempts → block account

```python
correct_username = "admin"
correct_password = "1234"

attempts = 3

while attempts > 0:

    username = input("Enter username: ")
    password = input("Enter password: ")

    if username == correct_username and password == correct_password:
        print("Login Successful")
        break

    else:
        attempts -= 1
        print("Wrong Login")
        print("Attempts left:", attempts)


if attempts == 0:
    print("Account Blocked")
```
